```python
# median before filling nan
be_fillna_medi = df["horsepower"].median()
print(be_fillna_medi)

# median after filling nan with the most frequent value 
most_frequent = df["horsepower"].value_counts().max()
print(most_frequent)

df["horsepower"] = df["horsepower"].fillna(most_frequent)

af_fillna_medi = df["horsepower"].median()

print(af_fillna_medi)
```

```python
cars_from_asia = df[df["origin"]=="Asia"]
cars_from_asia.head()
```

```python
cars_from_asia = cars_from_asia[["vehicle_weight","model_year"]]
first_7columns = cars_from_asia.head(7)
```

```python
X = first_7columns.to_numpy()
print(X)
```

```python
Xtransport = X.transpose()
print(Xtransport)
```

```python
XTX = np.matmul(Xtransport,X)
print(XTX)
```

```python
inverse_XTX = np.linalg.inv(XTX)
print(inverse_XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = np.matmul(np.matmul(inverse_XTX,Xtransport),y)
print(w)
```
